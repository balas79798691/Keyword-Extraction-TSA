# Keyword Extraction Application

An application that identifies important keywords and phrases from unstructured text.

## 1. Install and Import

```python
!pip -q install yake

import re
import string
import pandas as pd
import matplotlib.pyplot as plt
import yake

print("Keyword Extraction application is ready.")
```

## 2. Text Preprocessing

```python
stop_words = {
    "the","a","an","and","or","but","is","are","was","were","to","of",
    "in","on","for","with","as","by","from","at","this","that","it","its"
}

def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    words = [w for w in text.split() if w not in stop_words]
    return " ".join(words)
```

## 3. Keyword Extraction Engine

```python
def extract_keywords(text, number=10):
    extractor = yake.KeywordExtractor(lan="en", n=2, dedupLim=0.7, top=number)
    keywords = extractor.extract_keywords(text)

    return pd.DataFrame(
        [(keyword, round(score, 5)) for keyword, score in keywords],
        columns=["Keyword", "Score"]
    )

sample = (
    "Artificial intelligence and machine learning are transforming modern "
    "software development. Developers use machine learning models to analyze "
    "data, automate repetitive tasks, build recommendation systems, and create "
    "intelligent applications. Cloud computing provides infrastructure for AI."
)

keywords = extract_keywords(sample, 10)
display(keywords)
```

## 4. Application

```python
def keyword_finder():
    print("=" * 60)
    print("                  KEYWORD FINDER")
    print("=" * 60)
    print("Enter a paragraph. Type END when finished.\n")

    lines = []
    while True:
        line = input()
        if line.strip().upper() == "END":
            break
        lines.append(line)

    text = " ".join(lines)

    if len(text.split()) < 10:
        print("Please enter a longer text.")
        return

    value = input("Number of keywords [10]: ")
    try:
        count = int(value) if value.strip() else 10
        count = max(1, min(count, 30))
    except ValueError:
        count = 10

    result = extract_keywords(text, count)

    print("\nIMPORTANT KEYWORDS")
    print("-" * 60)
    for i, row in result.iterrows():
        print(f"{i+1:02d}. {row['Keyword']}")

# Uncomment to run:
# keyword_finder()
```

## 5. Keyword Visualization

```python
plot_data = keywords.sort_values("Score", ascending=True)

plt.figure(figsize=(9,5))
plt.barh(plot_data["Keyword"], plot_data["Score"])
plt.title("Keyword Relevance Scores")
plt.xlabel("YAKE Score (lower is more relevant)")
plt.ylabel("Keyword")
plt.tight_layout()
plt.show()
```
